## step1
Here i combined all csv files then remvoed inf and nan vales then saved on Combine file

```python
import pandas as pd
import numpy as np
import os

# ─── INPUT & OUTPUT PATHS ───────────────────────────────────────────────
FOLDER_PATH = r"D:\Computer Engineering 2022\4-CE.Z Forth Year\second-semester-4.2\project\ML-NIDS-Android\03-dataset\01-raw"
OUTPUT_FILE = r"D:\Computer Engineering 2022\4-CE.Z Forth Year\second-semester-4.2\project\ML-NIDS-Android\03-dataset\02-processed\combined_clean.csv"
# ───────────────────────────────────────────────────────────────────────

def combine_and_clean(folder_path, output_file):
    # 1. Find all CSV files in the folder
    csv_files = [
        os.path.join(folder_path, f)
        for f in os.listdir(folder_path)
        if f.endswith(".csv")
    ]

    if not csv_files:
        print("No CSV files found in:", folder_path)
        return

    print(f"Found {len(csv_files)} CSV file(s):")
    for f in csv_files:
        print(f"  {os.path.basename(f)}")

    # 2. Load and merge all files
    dfs = []
    for f in csv_files:
        tmp = pd.read_csv(f, encoding='latin1')
        tmp.columns = tmp.columns.str.strip()
        print(f"  Loaded {os.path.basename(f)}: {tmp.shape}")
        dfs.append(tmp)

    df = pd.concat(dfs, ignore_index=True)
    print(f"\nCombined shape: {df.shape}")

    # 3. Replace infinity values with NaN, then drop all NaN rows
    print(f"Inf  values: {np.isinf(df.select_dtypes(include=np.number)).sum().sum()}")
    print(f"NaN  values: {df.isnull().sum().sum()}")

    df.replace([np.inf, -np.inf], np.nan, inplace=True)
    df.dropna(inplace=True)

    print(f"Shape after cleaning: {df.shape}")

    # 4. Save directly to the specific path
    df.to_csv(output_file, index=False)
    print(f"\nSaved → {output_file}")

    # 5. Quick label check
    if "Label" in df.columns:
        print("\nLabel distribution:")
        print(df["Label"].value_counts())


if __name__ == "__main__":
    combine_and_clean(FOLDER_PATH, OUTPUT_FILE)
```

```python
import pandas as pd
import numpy as np

# Path to your cleaned dataset
file_path = r"D:\Computer Engineering 2022\4-CE.Z Forth Year\second-semester-4.2\project\ML-NIDS-Android\03-dataset\02-processed\combined_clean.csv"

# Load dataset
df = pd.read_csv(file_path)

# 1. Check for NaN values
total_nan = df.isnull().sum().sum()
print("Total NaN values:", total_nan)

# 2. Check for infinity values
total_inf = np.isinf(df.select_dtypes(include=np.number)).sum().sum()
print("Total inf values:", total_inf)

# 3. Final confirmation
if total_nan == 0 and total_inf == 0:
    print("\n✅ Dataset is clean: no NaN or inf values remain.")
else:
    print("\n⚠️ Dataset still contains NaN or inf values.")
```
